```python
model = KeyedVectors.load_word2vec_format("/content/drive/MyDrive/MED-enem/cbow_s300.txt")
```

###Vetorização dos enunciados

```python
cleaned_data = pd.read_csv("/content/drive/MyDrive/MED-enem/cleaned_data.csv")
cleaned_data.head()
```

```python
remover_palavras=[] #vetor de palavras que não compoem o corpus do modelo
enunciado_limpo_novo = [] #lista de palavras presentes no enunciado depois de limpo
remover = [',', '.', ':', ';', '(', ')'] #lista de caracteres específicos a serem retirados
lista_enunciados = [] #lista de enunciados vetorizados

for indice in range(len(cleaned_data['enunciado_limpo'])):
  #print(indice)
  enunciado_vet =[] #lista para armazenar os vetores de todas as palavras de um enunciado
  enunciado = cleaned_data['enunciado_limpo'][indice].split()

  #remove as palavras que não estão no corpus
  for palavra in remover_palavras:
    if palavra in enunciado:
      enunciado = [x for x in enunciado if x != palavra]

  for palavra in enunciado:
    #remove os caracteres indesejados
    for i in remover:
      palavra = palavra.replace(i, "")

    if palavra.lower() in model:
      enunciado_vet.append(model[palavra.lower()])
    else:
      remover_palavras.append(palavra)

    #enunciado_vet.append(model[palavra.lower()])
  lista_enunciados.append(enunciado_vet)
```

```python
lista_textos = []

for i in range(len(cleaned_data)):
  #divide a string entre as 5 alternativas
  if pd.isna(cleaned_data['alternativas_limpo'][i]):
    print(i)
  else:
    lista_alternativas = cleaned_data['alternativas_limpo'][i].split(';')
    textos = []
    #divide a letra da alternativa do texto
    for item in lista_alternativas:
      item_dividido = item.split(':')
      textos.append(item_dividido[1])
  lista_textos.append(textos)
```
