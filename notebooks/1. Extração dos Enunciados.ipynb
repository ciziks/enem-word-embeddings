# 1. Extração dos Enunciados das Provas do ENEM

Este notebook apresenta o processo de extração textual das provas do ENEM, utilizando ferramentas e bibliotecas do Python, além de técnicas de OCR. Foram extraídas questões das provas de aplicação regular (Cor Azul) e das provas adaptadas para acessibilidade (Cor Laranja), conhecidas como provas LEDOR. Essas versões são destinadas a participantes com deficiência visual e trazem descrições detalhadas de imagens e gráficos, além de uma estrutura mais descritiva e padronizada.

No entanto, cada edição do exame apresentou particularidades que exigiram abordagens específicas. Assim, enquanto em alguns anos foi possível realizar a extração automática dos enunciados e alternativas — com posterior refinamento e revisão manual —, em outros anos foi necessário recorrer à extração manual.

Optamos por utilizar as provas LEDOR ao invés da aplicação regular nos anos disponíveis devido aos seguintes benefícios para a nossa tarefas de PLN e Predição:

- Enunciados mais completos e com menos dependência de elementos visuais;
- Formato mais limpo, facilitando a tokenização, análise semântica e a modelagem textual;
- Melhor desempenho na extração automática com bibliotecas como PyMuPDF, em conjunto com regex e revisão manual.


```python
# Importando as bibliotecas necessárias para extração de texto
import pymupdf
import re
import csv
import io
import pandas as pd
import pytesseract
from PIL import Image
```

## 1.1. Provas de Aplicação Regular


### 1.1.1. Extração Manual (2009 - 2014)


Devido à alta variação no formato dos PDFs ao longo desses anos — incluindo diferenças estruturais nas páginas, formatação dos textos e disposição das questões — optamos por não investir em um processo de automação individualizado para cada edição.

Em vez disso, realizamos a **extração manual** combinada com revisão textual simultânea, garantindo qualidade no conteúdo e integridade dos dados.


### 1.1.2. Extração Automática (2015 - 2023)


Nesta etapa, realizamos a extração dos enunciados das provas do ENEM entre 2015 e 2023, priorizando uma abordagem semiautomática:

- A extração foi realizada com a biblioteca *PyMuPDF* (*fitz*) do Python, que possibilitou acessar e estruturar os textos diretamente dos arquivos PDF com boa fidelidade;

- Após a extração automatizada, conduzimos uma revisão manual completa para corrigir eventuais erros de segmentação, formatação e garantir a integridade dos enunciados.

> Exceção: A prova regular de 2021 foi disponibilizada em um formato PDF criptografado e não extraível por bibliotecas padrão. Como alternativa, optamos por utilizar a prova do ENEM Digital 2021, que apresentava uma estrutura diferente, porém permitiu a extração do conteúdo com qualidade satisfatória após ajustes.


```python
# Definição de Variáveis de Input e Output
years = (2015, 2016, 2017, 2018, 2019, 2020, 2022, 2023)
```

```python
# Iterando pelos anos 2015-2023 (exceto 2021)
for year in years:

    # Definindo caminhos de input e output 
    input_path = f"../data/raw/provas/ENEM_{year}.pdf"
    output_path = f"../data/questions_extracted/enem_{year}.csv"

    # Extraindo texto da prova da página 1 até 15
    document = pymupdf.open(input_path)
    total_text = ""

    for page in document.pages(1, 15):
        text = page.get_text()
        total_text += text + "\n"

    # Remover o rodapé
    rodape_pattern = r"CN\s+•\s+2º\s+DIA\s+•\s+CADERNO\s+7\s+•\s+AZUL"
    total_text = re.sub(rodape_pattern, "", total_text)

    # Remover códigos do rodapé
    total_text = re.sub(r"\*\w+\*", "", total_text)

    # Remover linhas que contenham apenas dígitos (possíveis números de página)
    total_text = re.sub(r"\n\s*\d+\s*\n", "\n", total_text)
    total_text = re.sub(r"\n\s*\d+\s*$", "", total_text)

    # Dividindo o texto em blocos de questões
    question_blocks = re.split(r"QUESTÃO\s+", total_text)
    question_blocks = [block.strip() for block in question_blocks if block.strip()]

    results = []

    for block in question_blocks:
        # Extraindo número da questão
        m = re.match(r"(\d+)", block)
        if not m:
            continue
        num_quest = m.group(1)

        content = block[m.end() :].strip()

        # Separando enunciado e alternativas.
        parts = re.split(r"\nA\s*\n", content, maxsplit=1)
        if len(parts) == 2:
            enunciado = parts[0].strip()
            alt_text = "A\n" + parts[1].strip()
        else:
            enunciado = content
            alt_text = ""

        # Extraindo alternativas
        alt_regex = r"([A-E])\s+[A-E]\s+(.*?)(?=(?:\n[A-E]\s+[A-E]\s+)|\Z)"
        alt_raw = re.findall(alt_regex, alt_text, flags=re.DOTALL)

        # Formatando alternativas como "A: texto"
        alternativas = "; ".join(
            [f"{letter}: {text.strip()}" for letter, text in alt_raw]
        )

        results.append(
            {
                "numero_questao": num_quest,
                "enunciado": enunciado,
                "alternativas": alternativas,
            }
        )

    # Salvando em CSV
    with open(output_path, "w", newline="", encoding="utf-8") as csvfile:
        fieldnames = ["numero_questao", "enunciado", "alternativas"]
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        writer.writeheader()
        for row in results[1:]:
            writer.writerow(row)

    print(f"Extração de {year} finalizada")
```

### 1.1.3. Exceção: Enem Digital 2021

Em 2021, foi aplicada a edição digital do ENEM, que apresentou uma estrutura significativamente diferente da versão regular. Como, por alguma razão, a prova da aplicação regular foi disponibilizada em um formato PDF não extraível (com conteúdo criptografado), optamos por realizar a extração automática utilizando outro método no Enem Digital.


```python
input_path = f"../data/raw/provas/ENEM_2021.pdf"
output_path = f"../data/questions_extracted/enem_2021.csv"

# Extraindo texto do ENEM Digital 2021
document = pymupdf.open(input_path)
total_text = ""

for page in document.pages(5, 53):
    text = page.get_text()
    total_text += text + "\n"
```

```python
# Padrão do ENEM Digital
question_pattern = r"Questão\s+(\d+)\s+-\s+Ciências da Natureza e suas Tecnologias(.*?)(?=Questão\s+\d+\s+-\s+Ciências da Natureza e suas Tecnologias|$)"
question_matches = re.findall(question_pattern, total_text, flags=re.DOTALL)

results = []

# Padrão de formatação das alternativas do ENEM Digital
alt_regex = r"\n(?:\d+\n)*([A-E])\n([^\n]+)"

for num_quest, block in question_matches:
    content = block.strip()

    alt_raw = re.findall(alt_regex, content)

    enunciado = content
    alt_match = re.search(alt_regex, content)
    if alt_match:
        enunciado = content[: alt_match.start()].strip()

    # Formatando alternativas como "A: texto"
    alternativas = "; ".join([f"{letter}: {text.strip()}" for letter, text in alt_raw])

    results.append(
        {
            "numero_questao": num_quest,
            "enunciado": enunciado,
            "alternativas": alternativas,
        }
    )
```

```python
# Salvando em CSV
with open(output_path, "w", newline="", encoding="utf-8") as csvfile:
    fieldnames = ["numero_questao", "enunciado", "alternativas"]
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    for row in results[1:]:
        writer.writerow(row)

print("Extração Enem Digital 2021 Finalizada")
```

## 1.2. Provas Adaptadas LEDOR


### 1.2.1. Extração Automática (2018 - 2023)

Utilizamos prioritariamente as versões LEDOR das provas, que apresentam adaptações para acessibilidade e não incluem imagens nas questões:

- Como essas provas são mais textuais e acessíveis, o conteúdo pode ser extraído com maior fidelidade via bibliotecas como PyMuPDF.
- A ausência de imagens e a estrutura descritiva tornam a extração mais eficaz e o resultado mais completo para modelos de NLP e análise de questões.

Isso se mostrou vantajoso para o modelo, pois os enunciados são mais limpos e completos em texto, evitando perdas de contexto visual e facilitando o processo de tokenização e análise textual.


```python
years = (2018, 2019, 2020, 2022, 2023)

# Iterando pelos anos da Prova LEODOR
for year in years:

    # Definindo input e output paths
    input_path = f"../data/raw/provas/ENEM_{year}_LEDOR.pdf"
    output_path = f"../data/questions_extracted/enem_{year}_LEDOR.csv"

    # Extraindo texto do ENEM
    document = pymupdf.open(input_path)
    total_text = ""

    for page in document.pages(1, 15):
        text = page.get_text()
        total_text += text + "\n"

    # Remover o rodapé
    rodape_pattern = r"CN\s+•\s+2º\s+DIA\s+•\s+CADERNO\s+7\s+•\s+AZUL"
    total_text = re.sub(rodape_pattern, "", total_text)

    # Remover códigos do rodapé, por exemplo: *020325AZ5*
    total_text = re.sub(r"\*\w+\*", "", total_text)

    # Remover linhas que contenham apenas dígitos (possíveis números de página)
    total_text = re.sub(r"\n\s*\d+\s*\n", "\n", total_text)
    total_text = re.sub(r"\n\s*\d+\s*$", "", total_text)

    # Dividindo o texto em blocos de questões
    question_blocks = re.split(r"QUESTÃO\s+", total_text)
    question_blocks = [block.strip() for block in question_blocks if block.strip()]

    results = []

    for block in question_blocks:
        # Extraindo número da questão
        m = re.match(r"(\d+)", block)
        if not m:
            continue
        num_quest = m.group(1)

        content = block[m.end() :].strip()

        # Separando enunciado e alternativas.
        parts = re.split(r"\nA\s*\n", content, maxsplit=1)
        if len(parts) == 2:
            enunciado = parts[0].strip()
            alt_text = "A\n" + parts[1].strip()
        else:
            enunciado = content
            alt_text = ""

        # Extraindo alternativas
        alt_regex = r"([A-E])\s+[A-E]\s+(.*?)(?=(?:\n[A-E]\s+[A-E]\s+)|\Z)"
        alt_raw = re.findall(alt_regex, alt_text, flags=re.DOTALL)

        # Formatando alternativas como "A: texto"
        alternativas = "; ".join(
            [f"{letter}: {text.strip()}" for letter, text in alt_raw]
        )

        results.append(
            {
                "numero_questao": num_quest,
                "enunciado": enunciado,
                "alternativas": alternativas,
            }
        )

    # Salvando em CSV
    with open(output_path, "w", newline="", encoding="utf-8") as csvfile:
        fieldnames = ["numero_questao", "enunciado", "alternativas"]
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        writer.writeheader()
        for row in results[1:]:
            writer.writerow(row)

    print(f"Extração de {year} finalizada")
```

### 1.2.2. **Exceções**: Provas de 2017 e 2021

As edições de 2017 e 2021 exigiram uma abordagem diferente. Por não termos encontrado versões em texto acessível ou extraível diretamente, foi necessário realizar extração via OCR. Esse processo envolveu a digitalização dos arquivos PDF e posterior refinamento manual para garantir a integridade e precisão dos textos extraídos.


```python
years = (2017, 2021)

for year in years:
    input_path = f"../data/raw/provas/ENEM_{year}_LEDOR.pdf"
    output_path = f"../data/questions_extracted/enem_{year}_LEDOR.csv"

    doc = pymupdf.open(input_path)

    texto_total = ""

    for pagina in doc[1:16]:
        # Renderiza a página como imagem
        pix = pagina.get_pixmap(dpi=300)
        img = Image.open(io.BytesIO(pix.tobytes()))

        # OCR para extrair texto da imagem
        texto = pytesseract.image_to_string(img, lang="por")
        texto_total += texto + "\n"

    doc.close()

    # Remover o rodapé
    rodape_pattern = r"CN\s+•\s+2º\s+DIA\s+•\s+Caderno\s+11\s+•\s+LARANJA"
    full_text = re.sub(rodape_pattern, "", texto_total)

    # Remover códigos do rodapé, por exemplo: *020325AZ5*
    full_text = re.sub(r"\*\w+\*", "", full_text)

    # Remover linhas que contenham apenas dígitos (possíveis números de página)
    full_text = re.sub(r"\n\s*\d+\s*$", "", full_text)

    # Dividindo o texto em blocos de questões
    question_blocks = re.split(r"QUESTÃO\s+", full_text)
    question_blocks = [block.strip() for block in question_blocks if block.strip()]

    results = []

    for block in question_blocks:
        # Extraindo número da questão
        m = re.match(r"(\d+)", block)
        if not m:
            continue
        num_quest = m.group(1)

        content = block[m.end() :].strip()

        content = content.replace("\x03", "").replace("\x0f", "").replace("\x11", "")

        # Separando enunciado e alternativas.
        padrao = r"(\nA\t.*\nB\t.*\nC\t.*\nD\t.*\nE\t.*)"
        match = re.search(padrao, content, flags=re.DOTALL)
        if match:
            enunciado = content[: match.start()].strip()
            alt_text = match.group(0).strip()
        else:
            enunciado = content
            alt_text = ""

        # Extraindo alternativas
        alt_regex = r"([A-E])\s+(.*?)(?=\n[A-E]\s+|$)"
        alt_raw = re.findall(alt_regex, alt_text, flags=re.DOTALL)

        # Formatando alternativas como "A: texto"
        alternativas = "; ".join(
            [f"{letter}: {text.strip()}" for letter, text in alt_raw]
        )
        results.append(
            {
                "numero_questao": num_quest,
                "enunciado": enunciado,
                "alternativas": alternativas,
            }
        )

    with open(output_path, "w", newline="", encoding="utf-8") as csvfile:
        fieldnames = ["numero_questao", "enunciado", "alternativas"]
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        writer.writeheader()
        for row in results:
            writer.writerow(row)

    print(f"Extração de {year} finalizada")
```
